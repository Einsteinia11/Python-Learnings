## Q4 — Set operations: union, intersection, difference, symmetric difference
The most important part of sets. Know both symbol and method.

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# Predict each output
# 1:{1, 2, 3, 4, 5, 6, 7, 8}
# 2:{4, 5}
# 3:{1, 2, 3}
# 4:{6, 7, 8}
# 5:{1, 2, 3, 6, 7, 8}
# 6:
# 7:
# 8:

print(A | B)            # 1 — union (all elements)
print(A.union(B))       # 2 — same thing

print(A & B)            # 3 — intersection (common elements)
print(A.intersection(B)) # 4 — same thing

print(A - B)            # 5 — difference (in A but not B)
print(B - A)            # 6 — difference (in B but not A) — different!

print(A ^ B)            # 7 — symmetric difference (in either but not both)
print(A.symmetric_difference(B))  # 8 — same thing
```

```python
# Subset and superset checks
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}
C = {1, 2, 3}

# Predict True or False
# 1:True
# 2:True
# 3:True
# 4:False
# 5:True
# 6:True

print(A.issubset(B))      # 1 — is A a subset of B?
print(A <= B)             # 2 — same thing with operator
print(A < B)              # 3 — proper subset (A subset AND A != B)
print(A < C)              # 4 — proper subset of equal set?
print(B.issuperset(A))    # 5
print(A.isdisjoint({6, 7}))  # 6 — no common elements?
```

| Difference | - | `.difference()` |
| Symmetric difference | ^ | `.symmetric_difference()` |
| Subset | <= | `.issubset()` |

## Q5 — Update operations (in-place)
These modify the set instead of returning a new one.

```python
# Predict the output after each update
# 1: {1, 2, 3, 4, 5}
# 2: {3}
# 3: {1, 2}
# 4: {{1, 2, 4, 5}}

A = {1, 2, 3}
B = {3, 4, 5}

A |= B              # union update — adds B's elements to A
print(A)            # 1 

A = {1, 2, 3}
A &= B              # intersection update — keeps only common
print(A)            # 2

A = {1, 2, 3}
A -= B              # difference update — removes B's elements from A
print(A)            # 3

A = {1, 2, 3}
A ^= B              # symmetric difference update
print(A)            # 4
```

## Q7 — Practical uses of sets (interview scenarios)
Sets shine in specific situations. Know these patterns.

```python
# Pattern 1: Remove duplicates from a list
nums = [1, 3, 2, 3, 1, 4, 2, 5]

# Predict output
# 1: [1, 3, 2, 4, 5]
# 2: 5
unique = list(set(nums))
print(unique)              # 1 — order preserved?
print(len(unique))         # 2
```

```python
# Pattern 2: Fast membership check
# list `in` check is O(n), set `in` check is O(1)

allowed = {'admin', 'editor', 'viewer'}
role = 'editor'

# Predict output
# 1: True
# 2: False
print(role in allowed)         # 1
print('hacker' in allowed)     # 2
```

```python
# Pattern 3: Find common elements between two lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

# Predict output
# 1: {4, 5}
# 2: {1, 2, 3}
# 3: {1, 2, 3, 4, 5, 6, 7, 8} length = 8

common = set(list1) & set(list2)
print(common)                          # 1

only_in_list1 = set(list1) - set(list2)
print(only_in_list1)                   # 2

all_unique = set(list1) | set(list2)
print(len(all_unique))                 # 3
```

## Q8 — Set comprehension
Same as list comprehension — but produces a set.

```python
# Predict output and type
# 1: {1, 4, 9, 16, 25}
# 2: <class 'set'>
# 3: {2, 4, 6, 8}
# 4: {3, 4}

squares = {x**2 for x in range(6)}
print(squares)          # 1 — ordered?
print(type(squares))    # 2

# With condition
evens = {x for x in range(10) if x % 2 == 0}
print(evens)            # 3

# Duplicates get removed
last_digits = {x % 10 for x in [13, 23, 33, 14, 24]}
print(last_digits)      # 4 — how many unique last digits?
```

## Q9 — Frozen sets
Immutable version of set. Hashable — can be used as dict key.

```python
# Predict output or crash
# 1: {1, 2, 3}
# 2: <class 'frozenset'>
# 3: error
# 4: True
# 5: frozenset({2, 3})

fs = frozenset([1, 2, 3, 2, 1])
print(fs)              # 1
print(type(fs))        # 2

# fs.add(4)              # 3 — crash or ok?

# frozenset supports all read operations
print(2 in fs)         # 4
print(fs & frozenset([2, 3, 4]))  # 5 — set ops still work
```
