## Q2 — Reading multiple values: split() and map()
The most common input pattern in competitive programming and interviews.

````python
# Simulating: user typed '10 20 30' on one line
raw = '10 20 30'

# Predict what each line produces
# 1: ['10', '20', '30']
# 2: <class 'str'>
# 3: [10, 20, 30]
# 4: <class 'int'>

parts = raw.split()
print(parts)                          # 1 — what type? what values?
print(type(parts[0]))                 # 2 — string or int?

nums = list(map(int, raw.split()))
print(nums)                           # 3
print(type(nums[0]))                  # 4
````

## Q4 — f-string format specs (the ones you missed in the exam)
Learn these 10 specs and you can format anything.

````python
n = 42

# Predict the EXACT output — include spaces
# Use | as boundary marker to see alignment clearly

# 1:|42|
# 2:|          42|
# 3:|42          |
# 4:|     42     |
# 5:|000000000042|
# 6:||
# 7:
# 8:
# 9:
# 10:

print(f'|{n:10}|')     # 1 — default alignment for int
print(f'|{n:>10}|')    # 2 — right align
print(f'|{n:<10}|')    # 3 — left align
print(f'|{n:^10}|')    # 4 — center align
print(f'|{n:010}|')    # 5 — zero pad
print(f'{n:b}')        # 6 — binary
print(f'{n:o}')        # 7 — octal
print(f'{n:x}')        # 8 — hex lowercase
print(f'{n:X}')        # 9 — hex UPPERCASE
print(f'{1000000:,}')  # 10 — comma separator
````

````python
pi = 3.14159265

# Predict exact output
# 1: 3.14
# 2: 3.14159
# 3: |         3.14|
# 4: |         3.14|
# 5: |3.14         |
# 6: 3

print(f'{pi:.2f}')          # 1
print(f'{pi:.5f}')          # 2
print(f'|{pi:10.2f}|')      # 3 — width AND precision
print(f'|{pi:>10.2f}|')     # 4
print(f'|{pi:<10.2f}|')     # 5
print(f'{pi:.0f}')          # 6 — 0 decimal places, rounds?
````

````python
s = 'hi'

# Predict exact output
# 1: |hi|
# 2: |         hi|
# 3: |    hi     |
# 4: ****hi****

print(f'|{s:10}|')     # 1 — default string alignment (different from int!)
print(f'|{s:>10}|')    # 2
print(f'|{s:^10}|')    # 3
print(f'{s:*^10}')     # 4 — fill character
````

## Q5 — Three ways to format strings
Know all three. Know when to use each.

````python
name = 'Alice'
score = 95.678

# All three should produce identical output
# Predict the output (same for all 3):
# Output:

# Method 1: % formatting (old style)
print('Name: %s, Score: %.2f' % (name, score))

# Method 2: .format()
print('Name: {}, Score: {:.2f}'.format(name, score))

# Method 3: f-string (modern, preferred)
print(f'Name: {name}, Score: {score:.2f}')
````

````python
# .format() positional and keyword args — predict each
# 1:
# 2:
# 3:

print('{0} and {1}'.format('a', 'b'))        # 1
print('{1} and {0}'.format('a', 'b'))        # 2 — order swapped!
print('{x} + {y}'.format(x=10, y=20))       # 3 — keyword args
````

## Q8 — Reading input line by line in a loop
Common pattern in problems that read n lines of input.

````python
# Simulate: user gives n, then n numbers one per line
# inputs[0] = first line typed, inputs[1] = second line, etc.
inputs = ['3', '10', '20', '30']
idx = 0

def fake_input():
    global idx
    val = inputs[idx]
    idx += 1
    return val

# Predict what this prints
# Output:
#
#
#
#

n = int(fake_input())
total = 0
for _ in range(n):
    total += int(fake_input())
print(f'Sum: {total}')
print(f'Average: {total/n:.1f}')
````
