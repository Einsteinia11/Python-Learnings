## Q3 — `isinstance()` vs `type()` — the bool trap
One of the most common interview gotchas. Predict carefully.

````python
# Predict before running
#1 True
#2 True
#3 False
#4 True
#5 False
#6 True
#7 <class 'bool'>

print(isinstance(True, bool))    # 1
print(isinstance(True, int))     # 2  <-- most people get this wrong
print(isinstance(1, bool))       # 3  <-- tricky
print(type(True) == bool)        # 4
print(type(True) == int)         # 5
print(True == 1)                 # 6
print(bool)
````

**The rule:** `bool` is a **subclass** of `int`. So `isinstance(True, int)` is `True`.
But `type(True) == int` is `False` — `type()` checks the exact class, not the hierarchy.

## Q4 — `id()` and integer interning
Python caches small integers. Know the range and what it means. (3/4)

````python
# Predict True or False before running
# 1: True
# 2: False
# 3: True
# 4: False

a = 5
b = 5
print(a is b)          # 1 — small int, cached
print(a==b)

x = 1000
y = 1000
print(x is y)          # 2 — large int, not cached

print(id(1) == id(1))  # 3 — same object?
print(id(1000) == id(1000))  # 4 — same object?
print(id(x))
print(id(y))
print(id(x) == id(y)) 
print(id(100000) == id(100000)) 
````
