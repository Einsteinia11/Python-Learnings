## Q6 — Scope: LEGB rule
Local → Enclosing → Global → Built-in. Python looks in this order.

````python
# Predict each output
# 1:
# 2:
# 3:
# 4:

x = 'global'

def func():
    print(x)        # 1 — reads global x

func()

def func2():
    x = 'local'     # creates LOCAL x, doesn't touch global
    print(x)        # 2

func2()
print(x)            # 3 — global x unchanged?

def func3():
    print(x)        # 4 — what happens here?
    x = 'local'     # local assignment AFTER reading

func3()
````

````python
# global keyword
# Predict each output
# 1:
# 2:
# 3:

count = 0

def increment():
    global count     # tell Python: use the global count
    count += 1

increment()
increment()
increment()
print(count)         # 1

# Without global keyword
total = 0

def add(n):
    total += n       # no global keyword — crash?

add(5)               # 2
print(total)         # 3
````

````python
# nonlocal keyword — for nested functions
# Predict each output
# 1:
# 2:
# 3:

def outer():
    x = 10

    def inner():
        nonlocal x
        x += 5

    inner()
    print(x)         # 1 — did inner change outer's x?

outer()

def outer2():
    x = 10

    def inner2():
        x = 99       # no nonlocal — creates new local x

    inner2()
    print(x)         # 2 — outer's x unchanged?

outer2()
````

## Q7 — Lambda functions
Anonymous one-line functions. Used constantly with `sorted`, `map`, `filter`.

````python
# Predict each output
# 1:
# 2:
# 3:
# 4:
# 5:
# 6:

square = lambda x: x ** 2
print(square(5))              # 1
print(type(square))           # 2 — what type is a lambda?

add = lambda a, b: a + b
print(add(3, 4))              # 3

# Lambda with condition
is_even = lambda n: n % 2 == 0
print(is_even(4))             # 4
print(is_even(7))             # 5

# Lambda called immediately (IIFE)
print((lambda x, y: x * y)(3, 5))   # 6
````

````python
# Lambda with sorted, map, filter — predict each
# 1:
# 2:
# 3:
# 4:
# 5:

people = [('Alice', 30), ('Bob', 25), ('Charlie', 35), ('Diana', 28)]

# Sort by age
print(sorted(people, key=lambda p: p[1]))        # 1

# Sort by name length descending
print(sorted(people, key=lambda p: len(p[0]), reverse=True))  # 2

# map — apply function to every element
nums = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, nums))
print(squares)                                   # 3

# filter — keep elements where function returns True
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)                                     # 4

# What type does map() return before list()?
print(type(map(lambda x: x, nums)))              # 5
````

## Q8 — Functions are first-class objects
Functions can be passed as arguments, returned, stored in variables.

````python
# Predict each output
# 1:
# 2:
# 3:
# 4:
# 5:

def double(x): return x * 2
def triple(x): return x * 3

# Assigning function to variable
fn = double
print(fn(5))           # 1 — calling via variable
print(fn is double)    # 2 — same object?

# Passing function as argument
def apply(func, value):
    return func(value)

print(apply(double, 10))   # 3
print(apply(triple, 10))   # 4

# Storing functions in a list
ops = [double, triple, lambda x: x + 100]
print([f(5) for f in ops])  # 5
````

````python
# Returning a function from a function (closure)
# Predict each output
# 1:
# 2:
# 3:
# 4:

def multiplier(n):
    def inner(x):
        return x * n    # n comes from outer function
    return inner        # returns the function itself

double = multiplier(2)
triple = multiplier(3)

print(double(5))     # 1
print(triple(5))     # 2
print(type(double))  # 3
print(double(triple(2)))  # 4
````

## Q10 — Recursion
A function calling itself. Must have a base case or it runs forever.

````python
# Predict each output
# 1:
# 2:
# 3:
# 4:

def factorial(n):
    if n == 0 or n == 1:   # base case
        return 1
    return n * factorial(n - 1)

print(factorial(0))    # 1
print(factorial(1))    # 2
print(factorial(5))    # 3
print(factorial(10))   # 4
````

````python
# Trace the recursion — predict step by step
# factorial(4) calls:
# Step 1:
# Step 2:
# Step 3:
# Step 4 (base case):
# Unwinding:

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Predict
# 1:
# 2:
# 3:
# 4:

print(fibonacci(0))    # 1
print(fibonacci(1))    # 2
print(fibonacci(6))    # 3
print(fibonacci(10))   # 4
````
