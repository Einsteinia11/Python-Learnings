## Q3 — Adding, updating, and deleting keys

```python
# Predict each output or 'crash'
# 1: {'a': 1, 'b': 2, 'c': 3}
# 2: {'a': 99, 'b': 2, 'c': 3}
# 3: {'a': 99, 'c': 3}
# 4: {'a': 99, 'c': 3} - No error
# 5: 2
# 6: {'a': 1, 'c': 3}
# 7: pop the last value
# 8: error

d = {'a': 1, 'b': 2}

d['c'] = 3          # add new key
print(d)            # 1

d['a'] = 99         # update existing key
print(d)            # 2

del d['b']          # delete key
print(d)            # 3

# del d['z']          # delete non-existing key
print(d)            # 4 — crash?

d = {'a': 1, 'b': 2, 'c': 3}
val = d.pop('b')    # remove and return
print(val)          # 5
print(d)            # 6

val2 = d.pop('z', 'default')   # pop with default
print(val2)         # 7

# d.pop('z')          # pop non-existing, no default
print(d)            # 8 — crash?
```

Conclusion:
I confused list.pop() with dict.pop(). Lists pop by index (default = last item), while dictionaries pop by key and return the value for that key.

For dictionaries, always ask: "What if the key doesn't exist?" — get() returns a default/None, but del, d[key], and pop() (without default) raise KeyError.

```python
# popitem() and setdefault()
# Predict each
# 1:
# 2:
# 3:
# 4:
# 5:

d = {'a': 1, 'b': 2, 'c': 3}
item = d.popitem()     # removes and returns LAST inserted item
print(item)            # 1 — what type?
print(d)               # 2

d = {'a': 1, 'b': 2}
d.setdefault('a', 99)  # key exists — does it update?
print(d)               # 3

d.setdefault('c', 99)  # key doesn't exist — sets it
print(d)               # 4

print(d.setdefault('d', 0))  # 5 — what does setdefault() return?
```

## Q5 — Iterating over dictionaries
3 ways to iterate. Know what each gives you.

```python
d = {'name': 'Alice', 'age': 25, 'city': 'Delhi'}

# Predict the output of each loop

# Loop 1 — default iteration
# Output: 'name', 'age', 'city'
for item in d:
    print(item)

print('---')

# Loop 2 — iterate values
# Output: 'Alice', 25, 'Delhi'
for val in d.values():
    print(val)

print('---')

# Loop 3 — iterate key-value pairs
# Output: 'name' -> 'Alice', 'age' -> 25, 'city' -> 'Delhi'
for key, val in d.items():
    print(key, '->', val)
```

## Q6 — `update()` and merging dictionaries
3 ways to merge dicts. Know which ones overwrite duplicates.

```python
d1 = {'a': 1, 'b': 2}
d2 = {'b': 99, 'c': 3}   # 'b' exists in both!

# Predict each output
# 1: {'a': 1, 'b': 99, 'c': 3}
# 2: {'a': 1, 'b': 99, 'c': 3}
# 3: {'a': 1, 'b': 2}
# 4: {'a': 1, 'b': 2, 'c': 3}
# 5: {'a': 1, 'b': 2}

# Way 1 — update() modifies in place
d1.update(d2)
print(d1)          # 1 — what happened to 'b'?

# Way 2 — ** unpacking (Python 3.5+)
d1 = {'a': 1, 'b': 2}
d2 = {'b': 99, 'c': 3}
merged = {**d1, **d2}
print(merged)      # 2 — which 'b' wins?
print(d1)          # 3 — is d1 modified?

# Way 3 — | operator (Python 3.9+)
merged2 = d1 | d2
print(merged2)     # 4

d1 |= d2           # in-place merge
print(d1)          # 5
```

Same structure as list comprehension but produces a dict.

```python
# Predict each output
# 1: 1, 4, 9, 16, 25
# 2: 4, 16, 36, 64
# 3: 
# 4:
# 5:

# Basic
squares = {x: x**2 for x in range(1, 6)}
print(squares)       # 1

# With condition
even_sq = {x: x**2 for x in range(10) if x % 2 == 0}
print(even_sq)       # 2

# Swap keys and values
d = {'a': 1, 'b': 2, 'c': 3}
flipped = {v: k for k, v in d.items()}
print(flipped)       # 3

# From two lists
keys = ['name', 'age', 'city']
vals = ['Alice', 25, 'Delhi']
combined = {k: v for k, v in zip(keys, vals)}
print(combined)      # 4

# Filter — keep only items where value > 2
d = {'a': 1, 'b': 3, 'c': 2, 'd': 5}
filtered = {k: v for k, v in d.items() if v > 2}
print(filtered)      # 5
```

## Q10 — Copying dictionaries: the same shallow/deep trap as lists

```python
# Predict every output carefully
# 1: {'a': 1, 'b': 2, 'c': 3}
# 2: True
# 3: {'a': 1, 'b': [1, 2, 3]}
# 4: {'a': 1, 'b': [1, 2, 3, 99]}
# 5: {'a': 1, 'b': [1, 2, 3]}
# 6: {'a': 1, 'b': [1, 2, 3, 99]}

# Assignment — NOT a copy
d1 = {'a': 1, 'b': 2}
d2 = d1
d2['c'] = 3
print(d1)          # 1 — did d1 change?
print(d1 is d2)    # 2

# Shallow copy
d1 = {'a': 1, 'b': [1, 2, 3]}
d2 = d1.copy()
d2['a'] = 99       # change simple value
print(d1)          # 3 — did d1 change?

d2['b'].append(99) # change nested list
print(d1)          # 4 — did d1 change? THIS IS THE TRAP

# Deep copy
import copy
d1 = {'a': 1, 'b': [1, 2, 3]}
d2 = copy.deepcopy(d1)
d2['b'].append(99)
print(d1)          # 5 — safe now?
print(d2)          # 6
```

## Q11 — Common patterns: counting and grouping
The most frequent dict use-case in interviews.

```python
# Pattern 1 — frequency counter
# Predict the final output
# Output: {'apple': 3, 'banana': 2, 'cherry': 1}

words = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1
print(freq)
```

```python
# Pattern 2 — grouping items
# Predict the final output
# Output: 

students = [('Alice', 'A'), ('Bob', 'B'), ('Charlie', 'A'), ('Diana', 'B'), ('Eve', 'A')]
groups = {}
for name, grade in students:
    if grade not in groups:
        groups[grade] = []
    groups[grade].append(name)
print(groups)
```

```python
# Pattern 3 — invert a dictionary (swap keys and values)
# Predict the output
# Output: {1: 'a', 2: 'b', 3: 'c'}
# {1: 'a', 'b': 2, 3: 'c'}
original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}
print(inverted)

# What happens if values are NOT unique?
d = {'a': 1, 'b': 1, 'c': 2}
inverted2 = {v: k for k, v in d.items()}
print(inverted2)   # what key does 1 map to?
```

## Q12 — `defaultdict` and `Counter` from collections
These make dict patterns cleaner. Interviewers love them.

```python
from collections import defaultdict, Counter

# defaultdict — never raises KeyError, uses a default factory
# Predict each output
# 1:
# 2:
# 3:
# 4:

dd = defaultdict(int)   # default value is int() = 0
dd['a'] += 1
dd['a'] += 1
dd['b'] += 1
print(dd['a'])          # 1
print(dd['b'])          # 2
print(dd['z'])          # 3 — key doesn't exist, crash?
print(dict(dd))         # 4 — convert to regular dict
```

```python
# Counter — counts hashable objects automatically
# Predict each output
# 1: 
# 2:
# 3:
# 4:
# 5:

words = ['apple', 'banana', 'apple', 'cherry', 'banana', 'apple']
c = Counter(words)
print(c)                      # 1
print(c['apple'])             # 2
print(c['missing'])           # 3 — missing key, crash?
print(c.most_common(2))       # 4 — top 2

# Counter on a string
c2 = Counter('mississippi')
print(c2)                     # 5
```
