### 🧠 Theory Q2 — f-string format spec `f"{value:>10.3f}"`

**Your Answer:** `>` is telling that the value should be greater than 10.3f where f represents floating point and 3 is decimal points.

**Score: ❌ 0/1**

---
**✅ Correct Answer:**

The format spec `>10.3f` has 4 completely independent parts:

| Part | Meaning |
|------|---------|
| `>` | **Alignment** — right-align the value inside the field |
| `10` | **Field width** — total width of the output is 10 characters (padded with spaces) |
| `.3` | **Precision** — show 3 digits after the decimal point |
| `f` | **Type** — format as a fixed-point floating number |

So `f"{3.14159:>10.3f}"` → `'     3.142'` (right-aligned, 10 wide, 3 decimal places)

Alternatives: `<` = left-align, `^` = center, `0` = pad with zeros instead of spaces → `0>10.3f` gives `000003.142`

**Remark:** This is one of the most important f-string concepts. The `>` has nothing to do with 'greater than'. It means alignment direction. Think of it as an arrow pointing right.

**Score: ⚠️ 0.5/2**

### 💻 Code Q3 — XOR property to find odd-occurring number

**Your Answer:** *(blank — not attempted)*

**Score: ⬜ 0/2**

---
**✅ Correct Answer:**
```python
nums = [2, 3, 5, 3, 2, 5, 7, 7, 7]
result = 0
for n in nums:
    result ^= n      # XOR everything together
print(result)        # 7
```

**Why this works:**
- `n ^ n = 0` (any number XOR'd with itself = 0)
- `n ^ 0 = n` (any number XOR'd with 0 = itself)
- So all pairs cancel out to 0, leaving only the number that appears an odd number of times
- `2^2=0, 3^3=0, 5^5=0` — all cancel. `7^7^7 = 0^7 = 7`

### 💻 Code Q5 — While Loop Bug

**Your Answer:** The code keeps running forever because odd numbers keep growing.

**Score: ⚠️ 1/2**

---
**The actual bug:** `n = n / 2` uses `/` (float division) instead of `//` (integer division).

When `n=100`: `n = 100 / 2 = 50.0` (float). Now `50.0 % 2 == 0` is True, so `n = 25.0`. Then `25.0` is odd, so `n = 3 * 25.0 + 1 = 76.0`... Eventually `n = 1.0`. But the condition is `n != 1` — and `1.0 != 1` is **False** in Python (because `1.0 == 1`), so it actually stops. But the float conversion is still wrong practice.

The **real infinite loop risk**: when `n` becomes a float like `2.0`, then `2.0/2 = 1.0`, and `1.0 == 1` so it stops. But if floating point errors crept in and gave `0.9999...` instead of `1.0`, it would loop forever.

This is the **Collatz conjecture** sequence: always terminate at 1 (unproven for all numbers, but works in practice).

**✅ Fixed version:**
```python
n = 100
while n != 1:
    if n % 2 == 0:
        n = n // 2    # integer division — keeps n as int
    else:
        n = 3 * n + 1
print('Done')
```

**✅ Correct Answer:**
```python
def compress(s):
    if not s:
        return s
    result = ''
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:     # same as previous character
            count += 1
        else:
            result += s[i-1] + str(count)  # add char + count
            count = 1
    result += s[-1] + str(count)   # don't forget the last group
    return result if len(result) < len(s) else s

print(compress('aabcccdddd'))  # a2b1c3d4
print(compress('abc'))         # abc (compressed is longer)
```

**Key insight:** Walk through the string comparing each character to the previous one. When it changes, record the group.
